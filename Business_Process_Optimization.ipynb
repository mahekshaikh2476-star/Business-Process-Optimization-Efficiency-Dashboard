```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
```

```python
import random

np.random.seed(42)

process_names = [
    "Order Processing",
    "Invoice Approval",
    "Customer Support",
    "Inventory Management",
    "Recruitment",
    "Procurement",
    "Payroll Processing"
]

departments = [
    "Operations",
    "Finance",
    "HR",
    "Customer Service",
    "Supply Chain",
    "IT"
]

owners = [
    "John",
    "Mary",
    "David",
    "Sarah",
    "Michael",
    "Emma",
    "James"
]

data = []

for i in range(100):

    start_date = datetime(2025,1,1) + timedelta(days=np.random.randint(0,365))
    cycle_time = np.random.randint(2,24)

    end_date = start_date + timedelta(hours=cycle_time)

    task_volume = np.random.randint(100,1000)

    completed = np.random.randint(
        int(task_volume*0.7),
        task_volume
    )

    productivity = round((completed/task_volume)*100,2)

    errors = np.random.randint(0,20)

    rework = np.random.randint(0,10)

    process_cost = np.random.randint(1000,10000)

    complaints = np.random.randint(0,15)

    sla_achieved = np.random.randint(75,100)

    data.append([
        f"P{i+1:03}",
        random.choice(process_names),
        random.choice(departments),
        random.choice(owners),
        start_date.date(),
        end_date.date(),
        cycle_time,
        np.random.randint(4,20),
        np.random.randint(1,10),
        task_volume,
        completed,
        productivity,
        errors,
        rework,
        process_cost,
        complaints,
        95,
        sla_achieved,
        random.choice([
            "Completed",
            "Delayed",
            "In Progress"
        ]),
        random.choice([
            "Automation",
            "Training",
            "Process Redesign",
            "Resource Allocation"
        ])
    ])

columns = [
    "Process ID",
    "Process Name",
    "Department",
    "Process Owner",
    "Start Date",
    "End Date",
    "Cycle Time",
    "Standard Cycle Time",
    "Employee Assigned",
    "Task Volume",
    "Tasks Completed",
    "Productivity %",
    "Error Count",
    "Rework Count",
    "Process Cost",
    "Customer Complaints",
    "SLA Target",
    "SLA Achieved",
    "Process Status",
    "Improvement Opportunity"
]

df = pd.DataFrame(data, columns=columns)

df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
process_efficiency = (
    df["Standard Cycle Time"].mean()
    /
    df["Cycle Time"].mean()
) * 100

task_completion_rate = (
    df["Tasks Completed"].sum()
    /
    df["Task Volume"].sum()
) * 100

error_rate = (
    df["Error Count"].sum()
    /
    df["Task Volume"].sum()
) * 100

rework_rate = (
    df["Rework Count"].sum()
    /
    df["Task Volume"].sum()
) * 100

sla_rate = df["SLA Achieved"].mean()

avg_productivity = df["Productivity %"].mean()

total_cost = df["Process Cost"].sum()

print("Process Efficiency:", round(process_efficiency,2))
print("Task Completion:", round(task_completion_rate,2))
print("Error Rate:", round(error_rate,2))
print("Rework Rate:", round(rework_rate,2))
print("Average SLA:", round(sla_rate,2))
print("Productivity:", round(avg_productivity,2))
print("Total Cost:", total_cost)
```

```python
department_summary = df.groupby("Department").agg({
    "Process Cost":"sum",
    "Productivity %":"mean",
    "Error Count":"sum",
    "Tasks Completed":"sum"
})

department_summary
```

```python
process_summary = df.groupby("Process Name").agg({
    "Cycle Time":"mean",
    "Process Cost":"sum",
    "Error Count":"sum",
    "Productivity %":"mean"
})

process_summary
```

```python
bottlenecks = df[
    df["Cycle Time"] >
    df["Cycle Time"].mean()
]

bottlenecks[
    [
        "Process ID",
        "Process Name",
        "Department",
        "Cycle Time"
    ]
].head(20)
```
